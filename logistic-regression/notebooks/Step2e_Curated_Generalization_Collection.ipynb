```python
import subprocess
import sys
import os

RUN_ARGS = [
    sys.executable,
    'step2e_curated_generalization_collection.py',
    '--clear-path', os.path.join(PROJECT_ROOT, 'data/raw/CLEAR_corpus_final.xlsx'),
    '--out-dir', os.path.join(PROJECT_ROOT, 'outputs/curated_generalization_collection'),
    '--max-per-level', '1500',
    '--single-class-train-cap', '500',
    '--silver-cap-per-level', '1000',
    '--cefr-cap-per-level', '1000',
    '--cefr-doc-cap-per-level', '1500',
    '--synthetic-cap-per-level', '500',
    '--finrad-cap-per-level', '250',
]

print('Running:', ' '.join(RUN_ARGS))

# Execute the script in a separate python process to ensure clean module resolution
subprocess.run(RUN_ARGS, check=True)
```

```python
# Quick audit
import pandas as pd, os
OUT_DIR = os.path.join(PROJECT_ROOT, 'outputs/curated_generalization_collection')
for name in ['train', 'val', 'test', 'ood_race-middle', 'ood_race-high']:
    path = os.path.join(OUT_DIR, f'{name}.csv')
    if os.path.exists(path):
        df = pd.read_csv(path)
        print(name, 'rows=', len(df), 'labels=', df['education_level'].value_counts().to_dict())
summary_path = os.path.join(OUT_DIR, 'source_summary.csv')
if os.path.exists(summary_path):
    display(pd.read_csv(summary_path))
```
